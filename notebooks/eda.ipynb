### Data cleaning

```python
df.isnull().sum()
```

```python
df=df.drop(0,axis=0) #removed first row, as it is empty
```

```python
df=df.drop(["Personal_Social_Media_Fraud_Loss","Social_Media_Scam_Outcome"],axis=1) #dropping unwanted features
```

```python
df.shape #1509 records and 13 features
```

```python
df["Social_Media_Scam_Type"].value_counts()
```

```python
values_to_drop=[
    "Fake/impersonation account; Romance/relationship scam; Account takeover/hacking attempt",
    "Fake giveaway/prize; Fake job offer; Romance/relationship scam; Shopping/product scam",
    "Fake giveaway/prize; Fake job offer; Shopping/product scam",
    "Fake giveaway/prize; Investment/crypto scam; Romance/relationship scam; Shopping/product scam",
    "Fake giveaway/prize; Romance/relationship scam"
]
```

```python
df=df[~df["Social_Media_Scam_Type"].isin(values_to_drop)] #removing records that is not important for the prediction
```

```python
df["Social_Media_Scam_Type"].value_counts()
```

```python
df.shape #1504 records and 13 features
```

```python
df.head(3)
```

```python
df["Digital_Transactions_Per_Week"].value_counts()
```

```python
len(df[df["Digital_Transactions_Per_Week"].isnull()])
```

```python
mode_value = df["Digital_Transactions_Per_Week"].mode()[0]
```

```python
df["Digital_Transactions_Per_Week"] = (df["Digital_Transactions_Per_Week"].fillna(mode_value))
```

```python
print(df["Digital_Transactions_Per_Week"].value_counts(dropna=False))
```

```python
# Standardize inconsistent categories
mapping = {
    "more than 14 times/week": "More than 14 times / week",
    "never": "Never",
    "More than 30": "More than 14 times / week",
    "1-2 times": "1-3 times / week"
}

df["Digital_Transactions_Per_Week"] = (
    df["Digital_Transactions_Per_Week"]
    .str.strip()
    .replace(mapping)
)
```

```python
print(df["Digital_Transactions_Per_Week"].value_counts(dropna=False))
```

```python
df.isnull().sum()
```

```python
print(df["Transaction_Fraud_Loss"].value_counts())
```

```python
df["Transaction_Fraud_Loss"] = (
    df["Transaction_Fraud_Loss"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```

```python
print(df["Transaction_Fraud_Loss"].value_counts())
```

```python
len(df[df["Transaction_Fraud_Loss"].isnull()])
```

```python
mode_val=df["Transaction_Fraud_Loss"].mode()[0]
```

```python
df["Transaction_Fraud_Loss"] = (df["Transaction_Fraud_Loss"].fillna(mode_val))
```

```python
print(df["Transaction_Fraud_Loss"].value_counts(dropna=False))
```
