### Encoding

```python
from sklearn.preprocessing import FunctionTransformer

encoding_maps = {
    'Age_Group': {
        '18-25': 0,
        '26-35': 1,
        '36-45': 2,
        '46-55': 3,
        '56': 4
    },

    'Social_Media_Hours': {
        'Less than 1 hour': 0,
        '1-2 hours': 1,
        '2-4 hours': 2,
        '4-6 hours': 3,
        'More than 6 hours': 4,
        'More than 5 hours': 5
    },

    'Social_Media_Scam_Exposure': {
        'Never': 0,
        'Once': 1,
        '2-3 times': 2,
        '4-5 times': 3,
        'More than 5 times': 4
    },

    'Personal_Social_Media_Fraud_Loss': {
        'No': 0,
        'Yes': 1
    },

    'Transaction_Fraud_Loss': {
        'No': 0,
        'Yes': 1
    },

    'OTP_Password_Request_Exposure': {
        'Never': 0,
        'Once': 1,
        '2-3 times': 2,
        '4-5 times': 3,
        'More than 5 times': 4
    },

    'Suspicious_Link_Clicks': {
        'Never': 0,
        'Once': 1,
        '2-3 times': 2,
        '4-5 times': 3,
        'More than 5 times': 4
    }
}
```

```python
from sklearn.preprocessing import FunctionTransformer
import pandas as pd
import joblib

preprocessor = FunctionTransformer(
    func=pd.DataFrame.replace,
    kw_args={"to_replace": encoding_maps},
    validate=False,
    feature_names_out="one-to-one"
)
```

```python
df['Overall_Cybersecurity_Awareness'] = df['Overall_Cybersecurity_Awareness'].astype("str")
```

```python
mappings = {"1.0": 0, "2.0": 0, "3.0": 1, "4.0": 1, "5.0": 1}
df['Overall_Cybersecurity_Awareness'] = df['Overall_Cybersecurity_Awareness'].map(mappings)
```

```python
df["Overall_Cybersecurity_Awareness"].value_counts()
```

```python
Y = df["Overall_Cybersecurity_Awareness"].astype(int)

X = df.drop("Overall_Cybersecurity_Awareness",axis=1)

print("X shape:", X.shape)
print("Y shape:", Y.shape)

print(Y.value_counts().sort_index())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)

X_train_encoded = preprocessor.fit_transform(X_train)

X_test_encoded = preprocessor.transform(X_test)
```

```python
import joblib
import pandas as pd

# Save the fitted preprocessor
joblib.dump(preprocessor, "../models/preprocessor.pkl")

# Convert encoded arrays into DataFrames
feature_names = preprocessor.get_feature_names_out()

X_train_df = pd.DataFrame(
    X_train_encoded,
    columns=feature_names,
    index=X_train.index
)

X_test_df = pd.DataFrame(
    X_test_encoded,
    columns=feature_names,
    index=X_test.index
)

# Save datasets
X_train_df.to_csv("../data/X_train_encoded_new.csv", index=False)
X_test_df.to_csv("../data/X_test_encoded_new.csv", index=False)

Y_train.to_csv("../data/Y_train_new.csv", index=False)
Y_test.to_csv("../data/Y_test_new.csv", index=False)
```
