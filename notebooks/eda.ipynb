```python
target_mapping = {
    "very poor": 0,
    "poor": 1,
    "average": 2,
    "good": 3,
    "excellent": 4
}

Y = (
    df["Fraud_Identification_Ability"]
    .str.strip()
    .str.lower()
    .map(target_mapping)
)

X = df.drop(columns=["Fraud_Identification_Ability"])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

ordinal_features = [
    "Age_Group",
    "Social_Media_Hours",
    "Social_Media_Scam_Exposure",
    "Suspicious_Payment_UPI_QR_Exposure",
    "Suspicious_Link_Exposure",
    "OTP_Password_Request_Exposure",
    "Suspicious_Link_Clicks",
    "Security_Practices",
    "Digital_Transactions_Per_Week"
]

nominal_features = [
    "Social_Media_Scam_Type",
    "Transaction_Fraud_Loss"
]
```

```python
ordinal_categories = [
    ["18-25", "26-35", "36-45", "46-55", "56"],
    ["Less than 1 hour", "1-2 hours", "2-4 hours", "4-6 hours", "More than 6 hours"],
    ["Never", "Once", "2-3 times", "4-5 times", "More than 5 times"],
    ["Never", "Once", "2-3 times", "4-5 times", "More than 5 times"],
    ["Never", "Rarely", "Sometimes", "Often", "Very often"],
    ["Never", "Once", "2-3 times", "4-5 times", "More than 5 times"],
    ["Never", "Once", "2-3 times", "4-5 times", "More than 5 times"],
    ["Never", "Rarely", "Sometimes", "Often", "Always"],
    ["Never", "1-3 times / week", "4-7 times / week", "8-14 times / week", "More than 14 times / week", "1-3 times / day"]
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "ordinal",
            OrdinalEncoder(
                categories=ordinal_categories,
                handle_unknown="use_encoded_value",
                unknown_value=-1
            ),
            ordinal_features
        ),

        (
            "nominal",
            OneHotEncoder(handle_unknown="ignore"),
            nominal_features
        )
    ],
    remainder="passthrough"
)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)

X_train_encoded = preprocessor.fit_transform(X_train)

X_test_encoded = preprocessor.transform(X_test)
```

```python
import joblib
import pandas as pd

# Save the fitted preprocessor
joblib.dump(preprocessor, "../models/preprocessor.pkl")

# Convert encoded arrays into DataFrames
feature_names = preprocessor.get_feature_names_out()

X_train_df = pd.DataFrame(
    X_train_encoded,
    columns=feature_names,
    index=X_train.index
)

X_test_df = pd.DataFrame(
    X_test_encoded,
    columns=feature_names,
    index=X_test.index
)

# Save datasets
X_train_df.to_csv("../data/processed/X_train_encoded.csv", index=False)
X_test_df.to_csv("../data/processed/X_test_encoded.csv", index=False)

Y_train.to_csv("../data/processed/Y_train.csv", index=False)
Y_test.to_csv("../data/processed/Y_test.csv", index=False)
```
